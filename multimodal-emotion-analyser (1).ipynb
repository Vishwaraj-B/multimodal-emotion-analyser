```python
import cv2
import torch
import torchvision.transforms as transforms
from decord import VideoReader, cpu

# Keep this exactly the same
frame_transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.48145466, 0.4578275, 0.40821073],
                         std=[0.26862954, 0.26130258, 0.27577711])
])

# Replace extract_frames with decord version
def extract_frames(video_path, num_frames=8):
    try:
        vr = VideoReader(video_path, ctx=cpu(0))
        total_frames = len(vr)
        
        if total_frames == 0:
            return None
        
        indices = np.linspace(0, total_frames - 1, num_frames, dtype=int)
        frames = vr.get_batch(indices).asnumpy()  # (8, H, W, 3)
        
        result = []
        for frame in frames:
            result.append(frame_transform(frame))
        
        return torch.stack(result)  # (8, 3, 224, 224)
    
    except Exception as e:
        return None

# Test speed
import time
test_video = f'{TRAIN_VIDEOS}/dia0_utt0.mp4'
start = time.time()
frames = extract_frames(test_video)
elapsed = time.time() - start
print(f"Frames shape: {frames.shape}")
print(f"Time: {elapsed:.2f} seconds")
```

```python
import subprocess
import tempfile
import torchaudio
import os

def extract_audio(video_path, target_sr=16000):
    # Step 1: create a temporary file to store the extracted audio
    with tempfile.NamedTemporaryFile(suffix='.wav', delete=False) as tmp:
        tmp_path = tmp.name
    
    try:
        # Step 2: use ffmpeg to extract audio from the video
        subprocess.run([
            'ffmpeg',        # call ffmpeg
            '-i', video_path,  # input file
            '-ar', str(target_sr),  # set sample rate to 16000 Hz
            '-ac', '1',      # convert to mono (1 channel)
            '-y',            # overwrite temp file if it exists
            tmp_path         # output file
        ], capture_output=True)  # capture_output=True silences ffmpeg's console output
        
        # Step 3: load the wav file into a PyTorch tensor
        waveform, sr = torchaudio.load(tmp_path)
        # waveform shape is (1, samples) - 1 because mono
        # .squeeze(0) removes that first dimension → shape becomes (samples,)
        return waveform.squeeze(0)
    
    except Exception as e:
        print(f"Error: {e}")
        return None
    
    finally:
        # Step 4: always delete the temp file, even if something crashes
        if os.path.exists(tmp_path):
            os.unlink(tmp_path)

# Test it
audio = extract_audio(test_video)
print(f"Audio shape: {audio.shape}")
print(f"Duration: {audio.shape[0] / 16000:.2f} seconds")
print(f"Min value: {audio.min():.4f}")
print(f"Max value: {audio.max():.4f}")
```

```python
from torch.utils.data import Dataset, DataLoader

class MELDDataset(Dataset):
    def __init__(self, df, video_dir):
        # Store the dataframe and video directory path
        self.df = df.reset_index(drop=True)
        self.video_dir = video_dir
        self.max_audio_length = 128000  # 8 seconds at 16000 Hz
    
    def __len__(self):
        # How many samples total
        return len(self.df)
    
    def __getitem__(self, idx):
        # Get one row from dataframe
        row = self.df.iloc[idx]
        
        # Build video file path from dialogue and utterance id
        video_path = os.path.join(
            self.video_dir,
            f"dia{row['Dialogue_ID']}_utt{row['Utterance_ID']}.mp4"
        )
        
        # Extract frames (8, 3, 224, 224)
        frames = extract_frames(video_path)
        if frames is None:
            frames = torch.zeros(8, 3, 224, 224)
        
        # Extract audio and fix length
        audio = extract_audio(video_path)
        if audio is None:
            audio = torch.zeros(self.max_audio_length)
        elif len(audio) < self.max_audio_length:
            # Pad with zeros if too short
            padding = torch.zeros(self.max_audio_length - len(audio))
            audio = torch.cat([audio, padding])
        else:
            # Truncate if too long
            audio = audio[:self.max_audio_length]
        
        # Get transcript text
        text = clean_text(str(row['Utterance']))
        
        # Get emotion label
        label = int(row['label'])
        
        return frames, audio, text, label
```

```python
import torch
import torch.nn as nn

class EmotionFusionModel(nn.Module):
    def __init__(self, clip_model, wav2vec_model, bert_model, num_classes=7):
        super().__init__()
        
        # Store the three pretrained branch models
        self.clip = clip_model
        self.wav2vec = wav2vec_model
        self.bert = bert_model
        
        # Output dimensions of each branch
        clip_dim = 768    # CLIP vision outputs 512-dim vector
        audio_dim = 768   # Wav2Vec2 outputs 768-dim vector
        text_dim = 768    # BERT outputs 768-dim vector
        fusion_dim = 512  # common dimension we project everything to
        
        # Project each branch output to same fusion_dim
        # so cross attention can work on equal sized vectors
        self.visual_proj = nn.Linear(clip_dim, fusion_dim)
        self.audio_proj = nn.Linear(audio_dim, fusion_dim)
        self.text_proj = nn.Linear(text_dim, fusion_dim)
        
        # Cross modal transformer fusion layer
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=fusion_dim,
            nhead=8,          # 8 attention heads
            dim_feedforward=1024,
            dropout=0.1,
            batch_first=True
        )
        self.fusion_transformer = nn.TransformerEncoder(encoder_layer, num_layers=2)
        
        # Final classification head
        self.classifier = nn.Sequential(
            nn.Linear(fusion_dim * 3, 256),  # 3 because we have 3 modalities
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )
    
    def forward(self, frames, audio_input, text_input):
        batch_size = frames.shape[0]
        
        # --- Visual Branch ---
        # frames shape: (batch, 8, 3, 224, 224)
        # reshape to process all 8 frames at once through CLIP
        frames_flat = frames.view(batch_size * 8, 3, 224, 224)
        with torch.no_grad():
            visual_out = self.clip.vision_model(pixel_values=frames_flat)
        # get CLS token output → (batch*8, 512)
        visual_features = visual_out.pooler_output
        # reshape back and average across 8 frames → (batch, 512)
        visual_features = visual_features.view(batch_size, 8, -1).mean(dim=1)
        visual_features = self.visual_proj(visual_features)  # (batch, 512)
        
        # --- Audio Branch ---
        with torch.no_grad():
            audio_out = self.wav2vec(input_values=audio_input)
        # wav2vec outputs (batch, time_steps, 768)
        # average across time → (batch, 768)
        audio_features = audio_out.last_hidden_state.mean(dim=1)
        audio_features = self.audio_proj(audio_features)  # (batch, 512)
        
        # --- Text Branch ---
        with torch.no_grad():
            text_out = self.bert(
                input_ids=text_input['input_ids'],
                attention_mask=text_input['attention_mask']
            )
        # BERT CLS token → (batch, 768)
        text_features = text_out.pooler_output
        text_features = self.text_proj(text_features)  # (batch, 512)
        
        # --- Fusion ---
        # stack 3 modalities as sequence → (batch, 3, 512)
        fused = torch.stack([visual_features, audio_features, text_features], dim=1)
        # cross modal transformer lets each modality attend to others
        fused = self.fusion_transformer(fused)  # (batch, 3, 512)
        # flatten → (batch, 3*512) = (batch, 1536)
        fused = fused.view(batch_size, -1)
        
        # --- Classification ---
        output = self.classifier(fused)  # (batch, 7)
        return output
```

```python
model = EmotionFusionModel(clip_model, wav2vec_model, bert_model, num_classes=NUM_CLASSES)
model = model.to(DEVICE)
print("Model created successfully!")

# Count trainable parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
```

```python
# Freeze all pretrained branch parameters
for param in model.clip.parameters():
    param.requires_grad = False

for param in model.wav2vec.parameters():
    param.requires_grad = False

for param in model.bert.parameters():
    param.requires_grad = False

# Verify
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print(f"Frozen parameters: {total_params - trainable_params:,}")
```

```python
import torch.optim as optim

# Loss function — CrossEntropy with class weights to handle imbalance
criterion = nn.CrossEntropyLoss(weight=class_weights)

# Optimizer — only pass trainable parameters
optimizer = optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4,
    weight_decay=0.01
)

# Learning rate scheduler — reduces LR when validation loss stops improving
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='min',
    patience=2,
    factor=0.5
)

print("Optimizer and loss function ready!")
```

```python
from tqdm import tqdm

def train_epoch(model, loader, optimizer, criterion, device):
    model.train()  # set model to training mode
    total_loss = 0
    correct = 0
    total = 0
    
    for batch in tqdm(loader, desc="Training"):
        frames, audio, texts, labels = batch
        
        # Move to GPU
        frames = frames.to(device)
        audio = audio.to(device)
        labels = labels.to(device)
        
        # Tokenize text for BERT
        text_input = bert_tokenizer(
            list(texts),
            return_tensors='pt',
            padding=True,
            truncation=True,
            max_length=128
        ).to(device)
        
        # Zero gradients from previous batch
        optimizer.zero_grad()
        
        # Forward pass
        outputs = model(frames, audio, text_input)
        
        # Calculate loss
        loss = criterion(outputs, labels)
        
        # Backward pass
        loss.backward()
        
        # Clip gradients to prevent exploding gradients
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        
        # Update weights
        optimizer.step()
        
        # Track metrics
        total_loss += loss.item()
        predicted = outputs.argmax(dim=1)
        correct += (predicted == labels).sum().item()
        total += labels.size(0)
    
    avg_loss = total_loss / len(loader)
    accuracy = correct / total * 100
    return avg_loss, accuracy


def eval_epoch(model, loader, criterion, device):
    model.eval()  # set model to evaluation mode
    total_loss = 0
    correct = 0
    total = 0
    
    with torch.no_grad():  # no gradients needed for evaluation
        for batch in tqdm(loader, desc="Evaluating"):
            frames, audio, texts, labels = batch
            
            frames = frames.to(device)
            audio = audio.to(device)
            labels = labels.to(device)
            
            text_input = bert_tokenizer(
                list(texts),
                return_tensors='pt',
                padding=True,
                truncation=True,
                max_length=128
            ).to(device)
            
            outputs = model(frames, audio, text_input)
            loss = criterion(outputs, labels)
            
            total_loss += loss.item()
            predicted = outputs.argmax(dim=1)
            correct += (predicted == labels).sum().item()
            total += labels.size(0)
    
    avg_loss = total_loss / len(loader)
    accuracy = correct / total * 100
    return avg_loss, accuracy


print("Training functions ready!")
```

```python
import os

NUM_EPOCHS = 10
best_val_loss = float('inf')
best_model_path = '/kaggle/working/best_model.pt'

train_losses, val_losses = [], []
train_accs, val_accs = [], []

for epoch in range(NUM_EPOCHS):
    print(f"\nEpoch {epoch+1}/{NUM_EPOCHS}")
    print("-" * 40)
    
    # Train
    train_loss, train_acc = train_epoch(model, train_loader, optimizer, criterion, DEVICE)
    
    # Evaluate
    val_loss, val_acc = eval_epoch(model, dev_loader, criterion, DEVICE)
    
    # Scheduler step
    scheduler.step(val_loss)
    
    # Track history
    train_losses.append(train_loss)
    val_losses.append(val_loss)
    train_accs.append(train_acc)
    val_accs.append(val_acc)
    
    # Save best model
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        torch.save(model.state_dict(), best_model_path)
        print(f"New best model saved!")
    
    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
    print(f"LR: {optimizer.param_groups[0]['lr']:.2e}")

print("\nTraining complete!")
```
